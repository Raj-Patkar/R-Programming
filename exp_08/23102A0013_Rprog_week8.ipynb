# Operations on Lists

```r
list1 <- list(1, 2, 3)
list2 <- list("water", "juice", "lemonade")

list1
list2
```

# Merging Lists

```r
list12 <- c(list1, list2)

list12
```

# Lists to Vector

```r
unlist(list1)
unlist(list2)

mode(list1)
mode(unlist(list1))
```

# Appending Lists

```r
append(list1, 100)
append(list2, "coffee")
```

```r
append(list1, 100, after = 2)
append(list2, "coffee", after = 2)
```

# Removing from List

```r
list1[-2]
list2[-1]
```

# Extracting from List

```r
list1 <- list(1, 2, 3, 4, 5, 6)
list2 <- list("water", "juice", "lemonade", "tea", "coffee", "milk")

list1[2:4]
list1[c(1, 3, 5)]

list2[2:4]
list2[c(1, 3, 5)]
```

# Vector Indexing

```r
x <- 1:10

x
x[x > 5]
x[x %% 2 == 0]
x[x %% 2 == 1]
```

# Vector with Missing Observations

```r
x <- 1:10
x[5] <- NA

x

y <- x[!is.na(x)]

y
mean(x)
mean(y)
```

# Factors

```r
y <- c(1, 4, 3, 5, 4, 2, 4)

possible.dieface <- c(1, 2, 3, 4, 5, 6)
labels.dieface <- c("one", "two", "three", "four", "five", "six")

facy <- factor(
  y,
  levels = possible.dieface,
  labels = labels.dieface
)

facy
```

# Factor Levels

```r
x <- factor(
  c("lemonade", "lemonade", "juice", "lemonade", "water")
)

x
levels(x)
```

# as.factor

```r
x <- c(3, 4, 5, 6, 1, 2, 3, 3, 4, 4, 5, 6)

y <- as.factor(x)

y
levels(y)
```

# Class Function

```r
class(9)
class("9")
class(print)

x <- matrix(nrow = 2, ncol = 2, data = 1:4)
class(x)
```

# Unclass Function

```r
brands <- c("A", "A", "B", "B", "B", "B", "C")
brands_fac <- factor(brands)

brands_fac
unclass(brands_fac)
```

```r
colours <- c("blue", "green", "red")

colours[unclass(brands_fac)]
```

```r
x <- factor(
  c("lemonade", "lemonade", "juice", "lemonade", "water")
)

x
unclass(x)
```

# Specifying Factor Levels

```r
x <- factor(
  c("lemonade", "lemonade", "juice", "lemonade", "water"),
  levels = c("water", "juice", "lemonade")
)

x
unclass(x)
levels(x)
```

# Ordered Factor

```r
income <- ordered(
  c("high", "high", "low", "medium", "medium"),
  levels = c("low", "medium", "high")
)

income
unclass(income)
```

# Print Function

```r
print(sqrt(2))
print(sqrt(2), digits = 5)
print(sqrt(2), digits = 10)

print("apple")
print(c("apple", "banana"))
print(c("apple", "banana", 6, 10))
```

# Format Function

```r
print(format(0.5, digits = 10, nsmall = 15))
```

# Format with Width

```r
format(
  c("A", "BB", "CCC", "DDDD"),
  width = 7,
  justify = "centre"
)

format(
  c("A", "BB", "CCC", "DDDD"),
  width = 14,
  justify = "centre"
)
```

# Format with Justify

```r
x <- c("A", "BB", "CCC", "DDDD")

format(x, width = 7, justify = "centre")
format(x, width = 7, justify = "left")
format(x, width = 7, justify = "right")
format(x, width = 7, justify = "none")
```
